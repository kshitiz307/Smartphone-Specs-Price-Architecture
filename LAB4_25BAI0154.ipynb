```python
import numpy as np
import pandas as pd
```

```python


df = pd.read_csv('/content/fraudTest.csv')
target_col = 'amt'

print("--- PART 1: DISCRETIZATION AND BINNING ---")

bins = [0, 50, 200, 1000, 50000]
category = pd.cut(df[target_col], bins)

print("\nBinned Intervals (First 5 Rows):")
print(category.head())

print("\nValue Counts per Bin (Updated Pandas Syntax):")
print(pd.Series(category).value_counts())

bin_names = ['Low Amount', 'Average Amount', 'High Amount', 'Extreme Amount']
labeled_category = pd.cut(df[target_col], bins, labels=bin_names)

print("\nLabeled Bins (First 5 Rows):")
print(labeled_category.head())

equal_length_bins = pd.cut(df[target_col], 5, precision=2)
print("\nEqual-Length Bins (5 uniform windows):")
print(pd.Series(equal_length_bins).value_counts())

category_quantiles = pd.qcut(df[target_col], 4, duplicates='drop')
print("\nBalanced Quantile Bins (Equal data volume per bin):")
print(pd.Series(category_quantiles).value_counts())


print("\n--- PART 2: DATA TRANSFORMATION ---")

min_val = df[target_col].min()
shift = abs(min_val) + 1 if min_val <= 0 else 0

df['Transformed_Amount'] = np.log(df[target_col] + shift)
print("Amount feature successfully normalized using natural logarithms.")
print(df[[target_col, 'Transformed_Amount']].head())


print("\n--- PART 3: OUTLIER DETECTION AND FILTERING ---")

TotalTransaction = df[target_col]

high_value_filter = TotalTransaction[np.abs(TotalTransaction) > 3000]
print(f"\nTransactions exceeding 3,000 (Detected matches: {len(high_value_filter)}):")
print(high_value_filter.head())

extreme_threshold = 10000
extreme_outliers_df = df[np.abs(TotalTransaction) > extreme_threshold]

print(f"\nRows where target value is strictly greater than {extreme_threshold}:")
print(f"Total Rows Found: {len(extreme_outliers_df)}")
if not extreme_outliers_df.empty:
    print(extreme_outliers_df[[target_col, 'trans_date_trans_time', 'category', 'merchant']].head())
else:
    print(f"No transactions found strictly exceeding {extreme_threshold}.")
```
